# Experiment 1: Find-S Algorithm (Concept Learning)

**Course:** ITA0617 - Machine Learning

```python
def find_s(training_data):
    num_attributes = len(training_data[0]) - 1
    hypothesis = ["0"] * num_attributes

    for row in training_data:
        attributes, label = row[:-1], row[-1]
        if label == "Yes":
            if hypothesis[0] == "0":
                hypothesis = attributes.copy()
            else:
                for i in range(num_attributes):
                    if hypothesis[i] != attributes[i]:
                        hypothesis[i] = "?"

    return hypothesis

# Attributes: [Sky, AirTemp, Humidity, Wind, Water, Forecast, EnjoySport]
dataset = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"],
]

final_hypothesis = find_s(dataset)
print("Find-S Algorithm")
print("Training Examples:", len(dataset))
print("Final Maximally Specific Hypothesis:", final_hypothesis)
```
