# Experiment 13: Car Price Prediction using Linear Regression

**Student:** Bhuvanesh H (192524027)  
**Course:** ITA0617 - Machine Learning

```python
from sklearn.linear_model import LinearRegression

# Feature: Engine / Age index, Target: Car Price
features = [[1], [2], [3], [4]]
prices = [100000, 200000, 300000, 400000]

model = LinearRegression()
model.fit(features, prices)

test_car = [[5]]
predicted_price = model.predict(test_car)

print("Car Price Prediction (Linear Regression)")
print("Input Feature:", test_car[0][0])
print("Predicted Car Price:", round(float(predicted_price[0]), 2))
```
