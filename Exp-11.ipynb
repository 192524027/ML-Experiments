# Experiment 11: Credit Score Classification using Decision Tree

**Student:** Bhuvanesh H (192524027)  
**Course:** ITA0617 - Machine Learning

```python
from sklearn.tree import DecisionTreeClassifier

# Credit scores and corresponding credit rating labels
credit_scores = [[500], [600], [700], [800]]
credit_ratings = ["Poor", "Fair", "Good", "Excellent"]

model = DecisionTreeClassifier(random_state=0)
model.fit(credit_scores, credit_ratings)

applicant_score = [[750]]
predicted_rating = model.predict(applicant_score)

print("Credit Score Classification")
print("Applicant Credit Score:", applicant_score[0][0])
print("Predicted Credit Rating:", predicted_rating[0])
```
