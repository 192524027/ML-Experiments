# Experiment 2: Candidate-Elimination Algorithm

```python
def is_consistent(hypothesis, sample):
    return all(h == "?" or h == s for h, s in zip(hypothesis, sample))


def candidate_elimination(training_data):
    num_attrs = len(training_data[0]) - 1
    specific_h = ["0"] * num_attrs
    general_h = [["?"] * num_attrs]

    for step, row in enumerate(training_data, start=1):
        sample, label = row[:-1], row[-1]

        if label == "Yes":
            for i in range(num_attrs):
                if specific_h[i] == "0":
                    specific_h[i] = sample[i]
                elif specific_h[i] != sample[i]:
                    specific_h[i] = "?"
            general_h = [g for g in general_h if is_consistent(g, sample)]
        else:
            new_general = []
            for g in general_h:
                if is_consistent(g, sample):
                    for i in range(num_attrs):
                        if g[i] == "?" and specific_h[i] != "?" and specific_h[i] != sample[i]:
                            candidate = g.copy()
                            candidate[i] = specific_h[i]
                            if candidate not in new_general:
                                new_general.append(candidate)
                else:
                    new_general.append(g)
            general_h = new_general

        print(f"Step {step} ({label}): S = {specific_h} | G = {general_h}")

    return specific_h, general_h

# Attributes: [Sky, AirTemp, Humidity, Wind, Water, Forecast, EnjoySport]
dataset = [
    ["Sunny", "Warm", "Normal", "Strong", "Warm", "Same", "Yes"],
    ["Sunny", "Warm", "High", "Strong", "Warm", "Same", "Yes"],
    ["Rainy", "Cold", "High", "Strong", "Warm", "Change", "No"],
    ["Sunny", "Warm", "High", "Strong", "Cool", "Change", "Yes"],
]

print("Candidate-Elimination Algorithm")
final_s, final_g = candidate_elimination(dataset)
print("Final Specific Boundary (S):", final_s)
print("Final General Boundary (G):", final_g)
```
