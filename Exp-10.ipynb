# Experiment 10: Expectation-Maximization (EM) Algorithm using Gaussian Mixture Model

**Student:** Bhuvanesh H (192524027)  
**Course:** ITA0617 - Machine Learning

```python
import numpy as np
from sklearn.mixture import GaussianMixture

# Unlabeled 1D observations from two distinct distributions
data = np.array([[1], [2], [3], [10], [11], [12]])

# Fit 2-component Gaussian Mixture Model via EM
em_model = GaussianMixture(n_components=2, random_state=0)
em_model.fit(data)

clusters = em_model.predict(data)

print("Expectation-Maximization (Gaussian Mixture Model)")
print("Input Data:", data.flatten().tolist())
print("Cluster Assignments:", clusters.tolist())
print("Estimated Component Means:", np.round(em_model.means_.flatten(), 2).tolist())
```
