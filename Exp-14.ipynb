# Experiment 14: House Price Prediction using Linear Regression

**Student:** Bhuvanesh H (192524027)  
**Course:** ITA0617 - Machine Learning

```python
from sklearn.linear_model import LinearRegression

# Feature: Area (sq. ft.), Target: Price (in Lakhs)
area_sqft = [[500], [1000], [1500], [2000]]
house_prices = [20, 40, 60, 80]

model = LinearRegression()
model.fit(area_sqft, house_prices)

new_house_area = [[2500]]
predicted_price = model.predict(new_house_area)

print("House Price Prediction (Linear Regression)")
print("House Area (sq. ft.):", new_house_area[0][0])
print("Predicted House Price (Lakhs):", round(float(predicted_price[0]), 2))
```
