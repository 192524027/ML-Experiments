# Experiment 20: Future Sales Prediction using Linear Regression

**Student:** Bhuvanesh H (192524027)  
**Course:** ITA0617 - Machine Learning

```python
from sklearn.linear_model import LinearRegression

# Advertising / Time index vs Sales volume
periods = [[1], [2], [3], [4]]
sales = [100, 200, 300, 400]

model = LinearRegression()
model.fit(periods, sales)

next_period = [[5]]
predicted_sales = model.predict(next_period)

print("Future Sales Prediction (Linear Regression)")
print("Target Period:", next_period[0][0])
print("Predicted Sales:", round(float(predicted_sales[0]), 2))
```
