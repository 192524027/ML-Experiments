```python
from sklearn.linear_model import LogisticRegression

# Study hours and Pass (1) / Fail (0) labels
study_hours = [[1], [2], [3], [6], [7], [8]]
results = [0, 0, 0, 1, 1, 1]

model = LogisticRegression()
model.fit(study_hours, results)

test_hours = [[5]]
prediction = model.predict(test_hours)
probability = model.predict_proba(test_hours)

print("Logistic Regression Classifier")
print("Test Input (Hours):", test_hours[0][0])
print("Predicted Class:", prediction[0])
print("Class Probabilities [Fail, Pass]:", probability[0].round(4).tolist())
```
