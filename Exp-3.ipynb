```python
from sklearn.tree import DecisionTreeClassifier, export_text

# Binary feature matrix and class labels
features = [[0, 0], [0, 1], [1, 0], [1, 1]]
labels = ["No", "No", "Yes", "Yes"]

# Initialize and fit ID3-style Decision Tree using Entropy
model = DecisionTreeClassifier(criterion="entropy", random_state=0)
model.fit(features, labels)

test_sample = [[1, 0]]
prediction = model.predict(test_sample)

print("Decision Tree Classifier (Entropy)")
print("Tree Structure:")
print(export_text(model, feature_names=["Feature_1", "Feature_2"]).strip())
print("Test Sample:", test_sample[0])
print("Predicted Class:", prediction[0])
```
